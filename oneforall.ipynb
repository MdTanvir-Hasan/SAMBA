```python
# Test device consistency more thoroughly
import torch
from config import ModelArgs, TrainingConfig
from models import SAMBA

print("=== Device Consistency Test ===")

# Check CUDA availability
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA device count: {torch.cuda.device_count()}")
    print(f"Current CUDA device: {torch.cuda.current_device()}")
    print(f"CUDA device name: {torch.cuda.get_device_name(0)}")

# Test with both CPU and CUDA
devices_to_test = ['cpu']
if torch.cuda.is_available():
    devices_to_test.append('cuda:0')

for device_name in devices_to_test:
    print(f"\n--- Testing on {device_name} ---")
    device = torch.device(device_name)
    
    # Create model
    model_args = ModelArgs(
        d_model=32,
        n_layer=2,
        vocab_size=10,
        seq_in=5,
        seq_out=1
    )
    
    model = SAMBA(
        model_args,
        hidden=32,
        inp=5,
        out=1,
        embed=10,
        cheb_k=3
    ).to(device)
    
    # Create input tensor on the same device
    input_tensor = torch.randn(2, 5, 10, device=device)
    
    # Test forward pass
    with torch.no_grad():
        output = model(input_tensor)
    
    print(f"✓ Model on {device_name}: Input {input_tensor.device} -> Output {output.device}")
    print(f"  Input shape: {input_tensor.shape}, Output shape: {output.shape}")

print("\n=== All device tests passed! ===")
```
